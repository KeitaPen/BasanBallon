```python
%matplotlib inline
import matplotlib.pyplot as plt

lines = []
x = []
y = []
with open("/content/drive/MyDrive/iris.data.txt","r") as f:
  for s in f:
    s = s.replace("\n", "")
    if s != "":
      line = s.split(",")
      for j in range(4):
        line[j] = float(line[j])
      lines.append(line)

for i in range(len(lines)):
  l = lines[i]
  x.append(l[0])
  y.append(l[1])

plt.scatter(x,y)
```
